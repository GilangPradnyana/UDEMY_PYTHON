```python
# Handling Missing value with new dataset
df = pd.read_csv('transactions_missing.csv')
df.head()
```

```python
# Checking the sum of dataset if there is missing value
df.isnull().sum()
```

```python
# df.head()
df['Sales_fillNA'] = df['Sales'].fillna(df['Sales'].median())
df
```

```python
# Renaming column
df = df.rename(columns={"Date" : "Sales Date"})
```

```python
# Changes data type
df["Value_new"] = df['Value'].fillna(df["Value"].mean()).astype(int)
df.head()
```

```python
df['New Value'] = df['Value'].apply(lambda x:x*2)
df.head()
```

```python
# Data aggregating and grouping
grouped_mean = df.groupby("Product")['Value'].mean()
grouped_mean.head()
```
