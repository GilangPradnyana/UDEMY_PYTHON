```python
# Create a Private Banking Account
class BankingAccount:
  def __init__(self, account_number, owner, balance) :
    self.__account_number = account_number
    self.__owner = owner
    self.__balance = balance

  # Getter Method
  def account_data(self):
    print("====== Account Info ======")
    print(f"Account Number : {self.__account_number}")
    print(f"Owner          : {self.__owner}")
    print(f"Balance        : {self.__balance}")
    print("==========================")

  # Deposit Money
  def deposit(self, ammount) :
    if ammount > 0:
      self.__balance += ammount
      print(f"Deposit successfull")
    else:
      print("Deposit ammount should be positive")

  # Withdraw Money
  def withdrawn(self, ammount) :
    if ammount <= 0 :
      print("Withdrawal ammount must be positive")
    elif ammount > self.__balance :
      print(f'Insufficient Funds!')
    else :
      self.__balance -= ammount
      print(f'Successfully Withdrawn {ammount}')


# Input Section
print("=== Create New Bank Account ===")
account_number = input("Enter Account Number : ")
owner = input("Input Name of the Account")
balance = int(input("Input balance : "))

# Create Account Object
account = BankingAccount(account_number, owner, balance)

# Menu Loop
while True :
  print("\n========== MENU ==========")
  print("1. Deposit")
  print("2. Withdraw")
  print("3. Check Balance")
  print("4. Exit")
  print("==========================")

  choice = input("Choose menu (1-4) : ")

  if choice == '1':
    ammount = float(input("Enter deposit ammount : "))
    account.deposit(ammount)

  elif choice == '2':
    ammount = float(input("Enter withdrawn ammount : "))
    account.withdrawn(ammount)

  elif choice == '3':
    account.account_data()

  elif choice == '4':
    print('\nThank you for using our banking system')
    break

  else :
    print('\nInvalid choice ! Please choose 1-4')
  
```

```python
# Inheritance Class

# Base Class
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

# Derive Class
class Employee(Person) : 
  def __init__(self, name, age, employee_id) :
    super().__init__(name, age)
    self.employee_id = employee_id

# Create an object of the derived class
emp1 = Employee("Gilang Pradnyana", 22, "EMP001")

print(f'Employee Name : {emp1.name}')
print(f'Employee Age : {emp1.age}')
print(f'Employee Employee ID : {emp1.employee_id}')
```

```python
# Override with __str__

# Base Class
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

# Derive Class
class Employee(Person) : 
  def __init__(self, name, age, employee_id) :
    super().__init__(name, age)
    self.employee_id = employee_id

  def __str__(self):
    return f'Employee name : {self.name}\n Employee Age : {self.age}\n ID : {self.employee_id}'

# Create an object of the derived class
emp1 = Employee("Gilang Pradnyana", 22, "EMP001")


print(emp1)
```

```python
# Static Method
import math

class MathOperations:
  @staticmethod
  def square_root(number):
    "Calculate and return the square root of a number"
    if number < 0 :
      return "Error : Cannot calculate square root of a negative number"
    return math.sqrt(number)

# ---- Calling the static method WITHOUT creating an object ----
print(MathOperations.square_root(16))
print(MathOperations.square_root(25))
print(MathOperations.square_root(2))
print(MathOperations.square_root(-9))   # handling negative input
```

```python
# Abstract
from abc import ABC, abstractmethod
import math

# Abstract
class Shape(ABC) :
  @abstractmethod
  def area(self):
    "Calculate the area of the shape"
    pass

# Derived class 1
class Circle(Shape):
  def __init__(self, radius):
    self.radius = radius

  def area(self):
    return math.pi * self.radius ** 2

# Derived class 2
class Square(Shape):
  def __init__(self, side):
    self.side = side

  def area(self):
    return self.side ** 2

# Testing
# Creating objects of the derived classes
circle = Circle(5)
square = Square(4)

print("Area of Circle:", circle.area())
print("Area of Square:", square.area())
```

```python
# Operator Overloading
class Vector:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  # Overloading the + operator
  def __add__(self, other) :
    # Create a new Vector by adding the corresponding components
    new_x = self.x + other.x
    new_y = self.y + other.y
    return Vector(new_x, new_y)

  def __str__(self):
    return f'Vector({self.x}, {self.y})'

  # ---- Testing ----
v1 = Vector(3, 4)
v2 = Vector(1, 2)

# Using the overloaded + operator
v3 = v1 + v2

print("v1 =", v1)
print("v2 =", v2)
print("v1 + v2 =", v3)
```

```python
# Class with custom exception
# Create a custom exception
class InssufficientBalanceError(Exception):
  "Raised when withdrawal amount is greater than available balance"
  pass

# BankAccount class that uses the custom exception
class BankAccount:
  def __init__(self, owner, balance=0):
    self.owner = owner
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount
    print(f'Deposited : {amount}, New Balance : {self.balance}')

  def withdraw(self, amount):
    if amount > self.balance:
      # Raise our custom exception
      raise InssufficientBalanceError(
        f'Cannot withdraw {amount}. Available balance is only {self.balance}'
      )
    self.balance -= amount
    print(f'Withdrawn {amount}. New Balance : {self.balance}')

    # ---- Testing ----
account = BankAccount("Andi", 1000)

print("--- Successful operations ---")
account.deposit(500)
account.withdraw(300)

print("\n--- Trying to withdraw too much ---")
try:
    account.withdraw(2000)          # This will raise the exception
except InssufficientBalanceError as e:
    print("Error:", e)
  
```
