```python
# Polymorphism
class Animal:
  def speak(self):
    return "Sound the Animmal"

class Dog(Animal):
  def speak(self):
    return 'Woof'

class Cat(Animal):
  def speak(self):
    return 'Meow'

def animal_speak(animal):
  print(animal.speak())

animal_speak(Dog())
animal_speak(Cat())
```

```python
class Bank:
  def __init__(self, owner, balance):
    self.__owner = owner
    self.__balance = balance

  def check_name(self):
    print("Hi ", self.__owner)

  def check_balance(self):
    print("My balance : ", self.__balance)


check = Bank("Gilang", 20)

check.check_name()
check.check_balance()
```
