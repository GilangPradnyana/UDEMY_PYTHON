```python
# Inheritance
class Animal():
  def __init__(self, name, species):
    self.name = name
    self.species = species

class Dog(Animal) :
  def __init__(self, name, species, breed) :
    # Call the parent class constructor
    super().__init__(name, species)
    self.breed = breed

  def show_info(self):
    print(f'Name : {self.name}, species : {self.species}, Breed : {self.breed}')

# Testing
dog1 = Dog("Buddy", "Canine", "Golden Retriver")

dog1.show_info()
print(dog1.name)
print(dog1.species)
print(dog1.breed)
```

```python
# Parent Class
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def __str__(self):
        return f"Animal: {self.name} ({self.species})"


# Child Class
class Dog(Animal):
    def __init__(self, name, species, breed):
        super().__init__(name, species)
        self.breed = breed

    # Overriding the __str__ method
    def __str__(self):
        return f"Dog: {self.name}, Species: {self.species}, Breed: {self.breed}"


# ---- Testing ----
dog1 = Dog("Buddy", "Canine", "Golden Retriever")


print(dog1)          # This will call the overridden __str__ method
```

```python
# MRO
# Base class 1
class Walker:
  def walk(self):
    print("Walker : i am walking normally")

class Runner:
  def run(self):
    print("Runner : i am running fast")

# Derived Class
class Athlete(Walker, Runner):
  def walk(self):
    print("Athlete : I am walking perfect form")
    #  call the walk method from the walker class using Super
    super().walk()

# Testing
athlete = Athlete()
athlete.walk()
```

```python
# Multiple Inheritance
class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age

  def introduce(self):
    print(f"Hi My name is {self.name}, iam {self.age} Years Old")


# Parent class 2
class SportPerson:
  def __init__(self, sport):
    self.sport = sport

  def play(self):
    print(f'I play {self.sport}')


# # Child class that inherits from both parents (Multiple Inheritance)
class Athlete(Person, SportPerson):
  def __init__(self,name,age,sport,training_hours):
    Person.__init__(self, name, age)
    SportPerson.__init__(self,sport)

    self.training_hours = training_hours

  # New method required by the assignment
  def training(self):
    print(f'{self.name} trains for {self.training_hours} hours')

# Object
athlete1 = Athlete(
  name = 'Budi',
  age = 22,
  sport = "Football",
  training_hours=15
)

# Call the methods
athlete1.introduce()   # from Person
athlete1.play()        # from SportsPerson
athlete1.training()       # new method we added
```

```python
# Diamond Problem in Multiple Inheritance
class A:
    def show(self):
        print("This is class A")


class B(A):
    def show(self):
        print("This is class B")


class C(A):
    def show(self):
        print("This is class C")


class D(B, C):          # Inherits from both B and C
    pass                # Does not override show()


# Create an object of class D
obj = D()

# Call the show method
obj.show()

# Check the Method Resolution Order
print("\nMethod Resolution Order (MRO):")
print(D.__mro__)
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Some generic animal sound")


class Cat(Animal):
    def speak(self):
        print("Meow!")


# Create objects
animal = Animal("Generic Animal")
cat = Cat("Mimi")


# Check instance types using isinstance()
print("Is animal an instance of Animal?", isinstance(animal, Animal))
print("Is animal an instance of Cat?", isinstance(animal, Cat))

print("Is cat an instance of Animal?", isinstance(cat, Animal))
print("Is cat an instance of Cat?", isinstance(cat, Cat))
```

```python
class Bird:
    def speak(self):
        print("Some generic bird sound")


class Parrot(Bird):
    def speak(self):
        print("Parrot says: Hello! Polly wants a cracker!")


class Penguin(Bird):
    def speak(self):
        print("Penguin says: Honk! Honk!")


# Create a list of Bird objects (polymorphism in action)
birds = [
    Parrot(),
    Penguin(),
    Bird(),
    Parrot()
]

# Call the same method on every object
print("Demonstrating Polymorphism:\n")
for bird in birds:
    bird.speak()
```

```python
class Device:
    def __init__(self, brand):
        self.brand = brand


class Phone(Device):                      # Single Inheritance
    def __init__(self, brand, model):
        Device.__init__(self, brand)      # Call parent constructor
        self.model = model


class Camera:
    def __init__(self, resolution):
        self.resolution = resolution


class Smartphone(Phone, Camera):          # Multiple Inheritance
    def __init__(self, brand, model, resolution):
        Phone.__init__(self, brand, model)    # Call Phone's constructor
        Camera.__init__(self, resolution)     # Call Camera's constructor


# Create an object of Smartphone
phone = Smartphone(
    brand="Samsung",
    model="Galaxy S24",
    resolution="200 MP"
)

# Print all attributes
print("Brand      :", phone.brand)
print("Model      :", phone.model)
print("Resolution :", phone.resolution)
```
