```python
def merge_files(file1, file2, output_file):
  with open(file1, "r") as f1:
    content1 = (f1.read())

  with open(file2, "r") as f2:
    content2 = (f2.read())

  merged_content = content1 + '\n' + content2 

  with open(output_file, 'w') as output :
    output.write(merged_content)

    print("Successfully merged into : ", output_file)
    print( '*' * 40)
    print('Content merged file : ')
    print(merged_content)

merge_files('destination.txt', 'example.txt', 'merged.txt')
```

```python
def split_file(input_file, lines_per_files) :
  with open(input_file, 'r') as file :
    lines = file.readlines()

  total_lines = len(lines)
  file_number = 1

  for i in range(0, total_lines, lines_per_files) :
    # get a chunk in line
    chunk = lines[i : i + lines_per_files]
    # create the new file name
    output_name = f'part_{file_number}.txt'
    # write the chunk into a new file
    with open(output_name, 'w') as output:
      output.writelines(chunk)

    print(f'Created : {output_name}')
    file_number += 1

# using it
split_file('destination.txt', 10)
```

```python
# Making a log file
from datetime import datetime

def write_log(message) :  
  # making a timestamp 
  timesstamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

  # Create the full log line
  log_line = f'[{timesstamp}] {message}'

  # Write to the log file (append mode)
  with open ('main.log', 'a') as log :
    log.write(log_line)

  print('Log Written Successfully')

def print_log(log_filename = 'main.log'):
  print("\n===== LOG FILE CONTENT =====")
  with open(log_filename, 'r') as read :
    for line in read:
      print(line, end="\n")

  print(("===== END OF LOG ====="))

write_log("Program started")
write_log("Processing data...")
write_log("Task completed")

print_log()



  
```

```python
def write_bin(filename, data):
    with open(filename, "wb") as f:
        f.write(data)
    print(f"Successfully wrote: {filename}")


def copy_binary_file(source_file, destination_file):
    with open(source_file, "rb") as bf:          # ← must be "rb"
        content = bf.read()                      # ← corrected from .rad()

    with open(destination_file, "wb") as bff:
        bff.write(content)

    print(f"Successfully copied '{source_file}' → '{destination_file}'")

def read_binary_file(bfile) :
    with open(bfile) as bf:
        print(bf.read())


# ========== Main Program ==========
binary_data = b"Hey it the new bin data \x00\x01\x02\x03\x04\x05 Hello Binary!\n"

# 1. Write the original binary file
write_bin("bin.bin", binary_data)

# 2. Copy it to a new file
copy_binary_file("bin.bin", "copy_bin.bin")

read_binary_file('copy_bin.bin')
```
