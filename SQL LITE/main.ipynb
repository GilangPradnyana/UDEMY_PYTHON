SQL (SQL LITE)

```python
import sqlite3
```

```python
# Connect to an SQL
connection = sqlite3.connect("example.db")
```

```python
cursor = connection.cursor()
```

```python
# Create a Table
cursor.execute('''
Create Table If Not Exists employees (
      id Integer Primary Key,
      name Text Not Null,
      age Integer,
      department text
      )
''')

# Commit the changes
connection.commit()
```

```python
cursor.execute('''
Select * from employees
''')
```

```python
# Insert the data in sqlite tabel
cursor.execute('''
Insert Into employees(name, age, department)
            values('Gilang', 32, 'Data Scientist')
''')

cursor.execute('''
Insert Into employees(name, age, department)
            values('Bob', 22, 'Web 3 Engineer')
''')

cursor.execute('''
Insert Into employees(name, age, department)
            values('Ray', 20, 'Banking')
''')

# Connection commit
connection.commit()
```

```python
# Query the data from sqlite
cursor.execute('Select * from Employees')
rows = cursor.fetchall()

# Print the data
for row in rows:
  print(row)
```
